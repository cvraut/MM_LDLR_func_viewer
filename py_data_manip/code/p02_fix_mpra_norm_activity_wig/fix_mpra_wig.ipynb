# Fix MPRA WIG

```python
import json, urllib.request
from pathlib import Path
import pandas as pd
import numpy as np

PROJECT_ROOT = Path("../").resolve()
OUT_DIR = PROJECT_ROOT / "tracks" / "new_data"
IN_DIR = PROJECT_ROOT / "tracks" / "raw_emmer"
```

```python
# load the wig data
wig_data = pd.read_table(f"{IN_DIR}/MPRA_normalized-activity.wig", comment="#", header=None, names=["pos", "value"], skiprows=1)
```

```python
wig_data
```

```python
wig_data.value.describe()
```

```python
# take the log2 of the values
wig_data["value2"] = np.log2(wig_data.value)
```

```python
wig_data["value2"].describe()
```

```python
# save the file to the output with the same name
wig_data[["pos","value2"]].to_csv(f"{OUT_DIR}/MPRA_normalized-activity.wig", index=False, header=False, sep="\t", na_rep="")
# figure out what the original first line was and add it back in
with open(f"{IN_DIR}/MPRA_normalized-activity.wig", "r") as f:
    first_line = f.readline().strip()
with open(f"{OUT_DIR}/MPRA_normalized-activity.wig", "r") as f:
    lines = f.readlines()
lines[0] = first_line + "\n"
with open(f"{OUT_DIR}/MPRA_normalized-activity.wig", "w+") as f:
    f.writelines(lines)
```
