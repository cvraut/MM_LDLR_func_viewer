# Get Fasta

```python
import json, urllib.request

chrom, start1, end1 = "chr19", 11_039_463, 11_163_820   # 1-based inclusive, as shown in the browser

# The API takes 0-based start, end exclusive
url = (f"https://api.genome.ucsc.edu/getData/sequence"
       f"?genome=hg38;chrom={chrom};start={start1 - 1};end={end1}")
dna = json.load(urllib.request.urlopen(url))["dna"]
assert len(dna) == end1 - start1 + 1, f"got {len(dna)} bases"

with open("../tracks/ucsc_raw/hg38_ldlr_region.fa", "w+", newline="\n") as f:
    f.write(f">{chrom}:{start1}-{end1}\n")
    for i in range(0, len(dna), 60):
        f.write(dna[i:i + 60] + "\n")
```

```python
# can copy the output to genome as raw, sequence is ok
import shutil
shutil.copy("../tracks/ucsc_raw/hg38_ldlr_region.fa", "../genome/hg38_ldlr_region.fa")
```
